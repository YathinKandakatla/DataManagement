# Research Perspective

## Research Questions

1. How are smoking-attributable expenditures, binge drinking prevalence, and high blood pressure prevalence distributed across U.S. states?

2. Which states stand out as having unusually high or low values within each public-health measure?

## Motivation

Smoking and excessive alcohol use are important behavioral health risk factors, while high blood pressure is an important cardiovascular health condition.

These three datasets were selected because they allow us to examine related public-health patterns at the state level.

The goal of this analysis is to describe state-level patterns within each dataset using descriptive statistics and basic data-management techniques. The analysis is exploratory and does not attempt to establish causal relationships.


## Hypothesis

Smoking-attributable expenditures, binge drinking prevalence, and high blood pressure prevalence will show substantial variation across U.S. states rather than being evenly distributed.

# Unit of Analysis

The main unit of analysis is the U.S. state.

Where year information is available, observations may also represent state-year combinations. State is used as the common geographic unit for describing patterns within the three datasets.


# Datasets

## Why These Three Datasets?

The Smoking-Attributable Expenditures dataset provides information about the economic burden associated with smoking.

The Current Binge Drinking among Adults dataset provides information about alcohol-related behavioral risk.

The High Blood Pressure Prevalence dataset provides information about an important cardiovascular health condition.

Together, these datasets provide three related public-health examples for practicing data import, descriptive statistics, filtering, regular expressions, grouping, and aggregation.

# Data Analysis

## 1. Smoking-Attributable Expenditures

### 1.1 Load and Inspect the Smoking Dataset

```python
import pandas as pd
import numpy as np
```

```python
smoking_url = "https://data.cdc.gov/api/views/ezab-8sq5/rows.csv?accessType=DOWNLOAD"
```

```python
smoking = pd.read_csv(smoking_url)
```

```python
smoking.head()
```

```python
# Check the dimensions of the raw dataset.
smoking.shape #(1560,19)
```

### 1.2 Examine the Dataset Structure

```python
smoking.columns
```

```python
smoking_clean = smoking[
    [
        "Year",
        "LocationAbbr",
        "LocationDesc",
        "TopicDesc",
        "MeasureDesc",
        "Variable",
        "Data_Value"
    ]
].copy()

```

```python
# Rename the columns so they are easier to read and use later in the analysis.

smoking_clean = smoking_clean.rename(
    columns={
        "Year": "Year",
        "LocationAbbr": "State_Code",
        "LocationDesc": "State",
        "TopicDesc": "Topic",
        "MeasureDesc": "Measure",
        "Variable": "Expenditure_Category",
        "Data_Value": "Expenditure_Value"
    }
)
smoking_clean.head()
```

```python
# Check the data types of the cleaned variables.
# This helps us identify whether numeric fields such as
# Expenditure_Value need to be converted before calculations.

smoking_clean.dtypes
```

```python
smoking_clean["Expenditure_Value"]
```

```python
# Check the data type of the expenditure value column.
# We do this before calculating statistics because the values need
# to be numeric for operations such as mean, median, minimum, and maximum.

smoking_clean["Expenditure_Value"].dtype
```

```python
smoking_clean["State"]
```

```python
smoking_clean["Expenditure_Value"].describe()
```

```python
# Calculate selected summary statistics separately.
# These values make it easier to compare the typical expenditure level
# and the range of smoking-attributable expenditures.

mean_exp = smoking_clean["Expenditure_Value"].mean()
median_exp = smoking_clean["Expenditure_Value"].median()
min_exp = smoking_clean["Expenditure_Value"].min()
max_exp = smoking_clean["Expenditure_Value"].max()
std_exp = smoking_clean["Expenditure_Value"].std()

mean_exp, median_exp, min_exp, max_exp, std_exp
```

### Interpretation of Descriptive Statistics

The descriptive statistics summarize the overall distribution of smoking-attributable expenditures in the dataset.

The mean represents the average expenditure across all observations, while the median shows the middle value of the distribution. The minimum and maximum values show the range of expenditures, and the standard deviation indicates how much the expenditure values vary across states, years, and expenditure categories.

A large difference between the mean and median may indicate that some observations have much higher expenditure values than others.

### Conditional Filtering

```python
# Filter the dataset to show only observations for New Jersey.
smoking_nj = smoking_clean[
    smoking_clean["State_Code"] == "NJ"
]

smoking_nj
```

```python
# Filter the dataset using two conditions at the same time.
# Here we keep only observations for New Jersey in the year 2009.
# This shows how multiple conditions can be combined using the & operator.

smoking_nj_2009 = smoking_clean[
    (smoking_clean["State_Code"] == "NJ") &
    (smoking_clean["Year"] == 2009)
]

smoking_nj_2009
```

```python
# Use .loc to filter rows and select only specific columns.
# Here we keep observations from the year 2009 and display only
# the state, expenditure category, and expenditure value.
# This demonstrates how .loc can perform row filtering and column selection together.

smoking_2009 = smoking_clean.loc[
    smoking_clean["Year"] == 2009,
    ["State", "Expenditure_Category", "Expenditure_Value"]
]

smoking_2009.head(10)
```

### Filtering with Regular Expressions

```python
# Use a regular expression to filter state names.
# This selects states whose names contain the words
# "New", "North", or "South".

smoking_regex = smoking_clean[
    smoking_clean["State"].str.contains(
        r"New|North|South",
        regex=True,
        na=False
    )
]

smoking_regex.head(15)
```

### Grouping and Aggregation

```python
# Group the dataset by year and calculate the average
# smoking-attributable expenditure value for each year.
# This gives us a simple summary of how expenditure values vary over time.

smoking_by_year = (
    smoking_clean
    .groupby("Year")["Expenditure_Value"]
    .mean()
)

smoking_by_year
```

```python
# Group the dataset by expenditure category.
# For each category, calculate the mean, median, minimum,
# maximum, and number of observations.
# This helps us compare the different types of smoking-related expenditures.

smoking_by_category = (
    smoking_clean
    .groupby("Expenditure_Category")
    .agg(
        Mean_Expenditure=("Expenditure_Value", "mean"),
        Median_Expenditure=("Expenditure_Value", "median"),
        Minimum_Expenditure=("Expenditure_Value", "min"),
        Maximum_Expenditure=("Expenditure_Value", "max"),
        Number_of_Observations=("Expenditure_Value", "count")
    )
)

smoking_by_category
```

```python
# Keep only observations where the expenditure category is "Total".
# This avoids mixing individual categories such as hospital,
# prescription drugs, or nursing home expenditures when comparing states.

smoking_total = smoking_clean[
    smoking_clean["Expenditure_Category"] == "Total"
].copy()

smoking_total.head()
```

```python
# Create a state-level summary of total smoking-attributable expenditures.
smoking_state_summary = (
    smoking_total
    .groupby(["State_Code", "State"])
    .agg(
        Average_Total_Expenditure=("Expenditure_Value", "mean"),
        Minimum_Total_Expenditure=("Expenditure_Value", "min"),
        Maximum_Total_Expenditure=("Expenditure_Value", "max")
    )
    .reset_index()
)
smoking_state_summary.head(10)
```

### Dataset 1 Summary

The Smoking-Attributable Expenditures dataset shows that smoking-related economic burden varies considerably across states, years, and expenditure categories.

The descriptive statistics show the overall distribution of expenditure values, while the filtering and grouping steps help identify state-level and category-level differences.

The state-level summary provides a concise view of how total smoking-attributable expenditures differ across states over the available years.


## 2. Current Binge Drinking among Adults

### 2.1 Load and Inspect the Alcohol Dataset

```python
alcohol_base_url = "https://chronicdata.cdc.gov/resource/dttw-5yxu.json"
```

```python
alcohol_url = (
    alcohol_base_url
    + "?$limit=500"
    + "&topic=Binge%20Drinking"
    + "&break_out=Overall"
)
```

```python
alcohol = pd.read_json(alcohol_url)
alcohol.head()
```

```python
alcohol.shape #(500,27)
```

```python
alcohol.columns
```

```python
alcohol["topic"].unique()
```

```python
alcohol["break_out"].unique()
```

```python
# Display the unique question wording used for the binge-drinking measure.
# This helps us understand exactly what the reported prevalence represents.

alcohol["question"].unique()
```

```python
# Display the available response categories.
# This helps us identify which rows correspond to the prevalence
# measure that should be used in the analysis.

alcohol["response"].unique()
```

```python
alcohol_clean = alcohol[
    [
        "year",
        "locationabbr",
        "locationdesc",
        "question",
        "response",
        "sample_size",
        "data_value"
    ]
].copy()

alcohol_clean.head()
```

```python
# Rename the columns so that they are easier to understand
alcohol_clean = alcohol_clean.rename(
    columns={
        "year": "Year",
        "locationabbr": "State_Code",
        "locationdesc": "State",
        "question": "Question",
        "response": "Response",
        "sample_size": "Sample_Size",
        "data_value": "Alcohol_Prevalence"
    }
)

alcohol_clean.head()
```

```python
# Check the data types of the selected variables.
alcohol_clean.dtypes
```

```python
# Keep only the "Yes" response for binge drinking.
# The dataset contains both "Yes" and "No" prevalence values for each state/year.
# Because these responses are complementary, averaging both would produce values
# near 50% and would not represent the actual prevalence of binge drinking.
# Keeping only "Yes" gives the estimated prevalence of adults classified as
# binge drinkers and makes the descriptive statistics meaningful.

alcohol_clean = alcohol_clean[
    alcohol_clean["Response"] == "Yes"
].copy()

alcohol_clean.head()
```

### Descriptive Statistics

```python
# Generate descriptive statistics for binge-drinking prevalence.
alcohol_clean["Alcohol_Prevalence"].describe()
```

```python
mean_alcohol = alcohol_clean["Alcohol_Prevalence"].mean()
median_alcohol = alcohol_clean["Alcohol_Prevalence"].median()
min_alcohol = alcohol_clean["Alcohol_Prevalence"].min()
max_alcohol = alcohol_clean["Alcohol_Prevalence"].max()
std_alcohol = alcohol_clean["Alcohol_Prevalence"].std()

mean_alcohol, median_alcohol, min_alcohol, max_alcohol, std_alcohol
```

### Interpretation of Descriptive Statistics

The descriptive statistics summarize the prevalence of binge drinking among adults across the observations included in the dataset.

The mean and median represent typical prevalence levels, while the minimum and maximum show the range across states and years. The standard deviation indicates how much binge-drinking prevalence varies across the observations.


### Conditional Filtering

```python
# Filter the dataset to show only observations for New Jersey.
# This demonstrates row selection using a single condition.

alcohol_nj = alcohol_clean[
    alcohol_clean["State_Code"] == "NJ"
]

alcohol_nj
```

```python
# Filter the dataset using two conditions.
# This keeps only New Jersey observations from the year 2023.

alcohol_nj_2023 = alcohol_clean[
    (alcohol_clean["State_Code"] == "NJ") &
    (alcohol_clean["Year"] == 2023)
]

alcohol_nj_2023
```

```python
# Select observations where binge-drinking prevalence is above 20%.
high_alcohol = alcohol_clean[
    alcohol_clean["Alcohol_Prevalence"] > 20
]

high_alcohol.head(10)
```

### Slicing with loc

```python
# Use .loc to filter rows while selecting only specific columns.
# Here we keep observations from 2023 and display only the
# state name, sample size, and alcohol prevalence.

alcohol_2023 = alcohol_clean.loc[
    alcohol_clean["Year"] == 2023,
    ["State", "Sample_Size", "Alcohol_Prevalence"]
]

alcohol_2023.head(10)
```

### Filtering with Regular Expressions

```python
# Use a regular expression to select states whose names contain
# "New", "North", or "South".
#
# This satisfies the assignment requirement to demonstrate
# text filtering using regular expressions.

alcohol_regex = alcohol_clean[
    alcohol_clean["State"].str.contains(
        r"New|North|South",
        case=False,
        regex=True,
        na=False
    )
]

alcohol_regex.head(15)
```

### Grouping and Aggregation

```python
# Group observations by year and calculate the average
# binge-drinking prevalence for each year.

alcohol_by_year = (
    alcohol_clean
    .groupby("Year")["Alcohol_Prevalence"]
    .mean()
)

alcohol_by_year
```

```python
# Group the dataset by state and calculate several summary statistics.
alcohol_state_summary = (
    alcohol_clean
    .groupby(["State_Code", "State"])
    .agg(
        Average_Alcohol_Prevalence=("Alcohol_Prevalence", "mean"),
        Minimum_Alcohol_Prevalence=("Alcohol_Prevalence", "min"),
        Maximum_Alcohol_Prevalence=("Alcohol_Prevalence", "max"),
        Number_of_Observations=("Alcohol_Prevalence", "count")
    )
    .reset_index()
)

alcohol_state_summary.head(10)
```

### Dataset 2 Summary

The binge-drinking dataset shows variation in alcohol prevalence across U.S. states and years.

Descriptive statistics summarize the overall distribution of binge-drinking prevalence, while conditional filtering demonstrates how individual states and higher-prevalence observations can be selected.

The state-level aggregation provides a concise summary of average binge-drinking prevalence across the available years.


## 3. High Blood Pressure Prevalence

```python
bp_url = "https://github.com/YathinKandakatla/DataManagement/raw/refs/heads/main/High%20BP.xlsx"
```

```python
bp = pd.read_excel(bp_url)
bp.head()
```

```python
bp.shape #(54, 27)
```

```python
bp.columns
```

```python
bp_clean = bp[
    [
        "Year",
        "Locationabbr",
        "Locationdesc",
        "Question",
        "Sample_Size",
        "Data_value",
        "Data_value_type"
    ]
].copy()

bp_clean.head()
```

```python
bp_clean = bp_clean.rename(
    columns={
        "Locationabbr": "State_Code",
        "Locationdesc": "State",
        "Sample_Size": "Sample_Size",
        "Data_value": "BP_Prevalence",
        "Data_value_type": "Prevalence_Type"
    }
)

bp_clean.head()
```

```python
bp_clean.dtypes
```

### Descriptive Statistics

```python
bp_clean["BP_Prevalence"].describe()
```

```python
mean_bp = bp_clean["BP_Prevalence"].mean()
median_bp = bp_clean["BP_Prevalence"].median()
min_bp = bp_clean["BP_Prevalence"].min()
max_bp = bp_clean["BP_Prevalence"].max()
std_bp = bp_clean["BP_Prevalence"].std()

mean_bp, median_bp, min_bp, max_bp, std_bp
```

### Interpretation of Descriptive Statistics

The descriptive statistics summarize how high blood pressure prevalence varies across the state-level observations in 2023.

The mean and median represent typical prevalence levels, while the minimum and maximum show the range across states. The standard deviation indicates how much high blood pressure prevalence varies across the observations.


### Conditional Filtering

```python
# Filter the dataset to show only New Jersey.
bp_nj = bp_clean[
    bp_clean["State_Code"] == "NJ"
]

bp_nj
```

```python
# Select states where high blood pressure prevalence is greater than 40%.
high_bp = bp_clean[
    bp_clean["BP_Prevalence"] > 40
]

high_bp
```

### Slicing with `loc`

```python
# Use .loc to filter rows and select only specific columns.
# Here we keep states with prevalence above 40% and display
# only the state name, sample size, and prevalence value.

bp_high_loc = bp_clean.loc[
    bp_clean["BP_Prevalence"] > 40,
    ["State", "Sample_Size", "BP_Prevalence"]
]

bp_high_loc
```

### Filtering with Regular Expressions

```python
# Use a regular expression to select state names containing
# "New", "North", or "South".
# This demonstrates text-based filtering using str.contains().

bp_regex = bp_clean[
    bp_clean["State"].str.contains(
        r"New|North|South",
        case=False,
        regex=True,
        na=False
    )
]

bp_regex
```

### Grouping and Aggregation

```python
# Group the dataset by prevalence type and calculate several
# descriptive statistics for each group.
bp_by_type = (
    bp_clean
    .groupby("Prevalence_Type")
    .agg(
        Mean_BP_Prevalence=("BP_Prevalence", "mean"),
        Median_BP_Prevalence=("BP_Prevalence", "median"),
        Minimum_BP_Prevalence=("BP_Prevalence", "min"),
        Maximum_BP_Prevalence=("BP_Prevalence", "max"),
        Number_of_States=("BP_Prevalence", "count")
    )
)

bp_by_type
```

```python
# Create a compact state-level summary.
# Because this dataset already contains one overall prevalence observation
# per state, we only keep the state identifiers and prevalence value.
# This provides a concise state-level view of high blood pressure prevalence.

bp_state_summary = bp_clean[
    [
        "State_Code",
        "State",
        "BP_Prevalence"
    ]
].copy()

bp_state_summary.head(10)
```

```python
# Sort the states from highest to lowest high blood pressure prevalence.
# This makes it easier to identify which states have relatively
# higher prevalence in the 2023 dataset.

bp_state_summary = bp_state_summary.sort_values(
    "BP_Prevalence",
    ascending=False
)

bp_state_summary.head(10)
```

### Dataset 3 Summary

The High Blood Pressure dataset contains state-level prevalence estimates for 2023.

The descriptive statistics summarize the overall distribution of high blood pressure prevalence, while conditional filtering and regular-expression filtering demonstrate how selected observations can be identified.

The state-level table provides a concise view of how high blood pressure prevalence differs across states in the 2023 subset.
